# PLS-SEM + CB-SEM analysis in Google Colab

**Step 1** – run the cell below and select **both** files: `pls_sem_analysis.py` and your data file `1.xlsx` (you can also add `simulate_data.py` for test data).

**Step 2** – run the analysis cell. Missing libraries (semopy, plotly, python-docx, ...) are installed automatically.

**Step 3** – the results (`PLS_SEM_Results.zip`) download automatically; key tables and figures are shown below the cell.

```python
# Step 1: upload pls_sem_analysis.py and 1.xlsx
import os, re
from google.colab import files
uploaded = files.upload()
# browsers rename repeated downloads to e.g. 'pls_sem_analysis (1).py' - restore the original names
for name in list(uploaded):
    clean = re.sub(r'\s*\(\d+\)(?=\.[^.]+$)', '', name)
    if clean != name:
        if os.path.exists(clean):
            os.remove(clean)
        os.rename(name, clean)
        print(f'Renamed {name!r} -> {clean!r}')
print('Files ready:', sorted(f for f in os.listdir('.') if f.endswith(('.py', '.xlsx', '.csv'))))
```
